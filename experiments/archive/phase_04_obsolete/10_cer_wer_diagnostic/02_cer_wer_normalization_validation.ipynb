```python
import sys
from pathlib import Path

# Add root workspace to sys.path
root_dir = Path.cwd()
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))

from src.document_processing.ocr_evaluation import compute_cer, compute_wer, normalize_text, evaluate_batch
print('Successfully imported ocr_evaluation utilities.')
```

```python
# Demonstration on benchmark sample pairs showing raw vs. normalized CER/WER impact
sample_pairs = [
    ('Cataract, Senile', 'cataract senile'),
    ('[✓] Married', '[✓] married'),
    ('0626-02', '624-0'),
    ('PHP 5,000.00', 'PHP 5000.00')
]

for ref, hyp in sample_pairs:
    raw_cer = compute_cer(ref, hyp, mode='raw')
    norm_cer = compute_cer(ref, hyp, mode='normalized')
    raw_wer = compute_wer(ref, hyp, mode='raw')
    norm_wer = compute_wer(ref, hyp, mode='normalized')
    print(f'Ref: {ref!r} | Hyp: {hyp!r}')
    print(f'  Raw -> CER: {raw_cer:.4f}, WER: {raw_wer:.4f}')
    print(f'  Norm -> CER: {norm_cer:.4f}, WER: {norm_wer:.4f}')
```
