```python
from pathlib import Path
import json
import pandas as pd

dict_root = Path('data/external/paddleocr/dictionaries')
char_dict_path = dict_root / 'character' / 'character_dictionary.txt'
domain_dict_path = dict_root / 'domain' / 'ofphil_domain_dictionary.csv'
abbrev_path = dict_root / 'domain' / 'approved_abbreviations.csv'

if char_dict_path.exists():
    with open(char_dict_path, 'r', encoding='utf-8') as f:
        chars = [line.strip('\r\n') for line in f if line.strip('\r\n')]
    print(f'Character dictionary loaded: {len(chars)} characters.')
else:
    print('Character dictionary not found.')
```

```python
if domain_dict_path.exists():
    df_domain = pd.read_csv(domain_dict_path)
    print(f'Domain dictionary loaded: {len(df_domain)} rows.')
    print(df_domain[['canonical_term', 'category', 'language', 'verification_status']].head())

if abbrev_path.exists():
    df_abbrev = pd.read_csv(abbrev_path)
    print(f'Approved abbreviations loaded: {len(df_abbrev)} abbreviations.')
    print(df_abbrev.head())
```
