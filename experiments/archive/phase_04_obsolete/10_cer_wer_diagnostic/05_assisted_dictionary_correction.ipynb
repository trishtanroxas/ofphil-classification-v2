```python
import sys
from pathlib import Path
root_dir = Path.cwd()
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))

from src.document_processing.ocr_correction import AssistedDomainCorrector
corrector = AssistedDomainCorrector()
print('AssistedDomainCorrector successfully initialized.')
print(f'Vocabulary size: {len(corrector.vocabulary)} terms.')
print(f'Approved abbreviations: {len(corrector.abbreviations)} entries.')
```

```python
# Run correction across test cases
test_cases_path = Path('data/external/paddleocr/dictionaries/evaluation/dictionary_test_cases.json')
if test_cases_path.exists():
    import json
    with open(test_cases_path, 'r', encoding='utf-8') as f:
        tc_data = json.load(f)
    cases = tc_data.get('test_cases', [])
    for tc in cases:
        inp = tc['source_text']
        mode = 'conservative'
        res = corrector.correct_text(inp, confidence=0.75, field_name=tc.get('category'), mode=mode)
        print(f'ID: {tc["test_id"]} | Input: {inp!r} -> Output: {res["output_text"]!r} | Protected: {res["is_protected"]}')
```
