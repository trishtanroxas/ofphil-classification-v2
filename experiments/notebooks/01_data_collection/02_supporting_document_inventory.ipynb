### Description & Purpose
Inventory physical and digitized supporting documents (e.g., Certificate of Indigency, Social Case Study, Medical Abstracts, Valid IDs), cataloging file types, resolutions, and availability for linkage.


### Research Objectives

1. Catalog supporting document files by type, format (PDF, JPG, PNG, TIFF), resolution, and file size.
2. Audit file organization across document repositories (`data/raw/documents/`).
3. Establish baseline document metadata without loading or exposing raw document contents.
4. Prepare a mapping inventory linking anonymized document IDs to patient profile keys.


### Inputs & Expected Outputs

**Required Inputs:**
- `Document storage directories (`data/raw/documents/samples/`, `data/raw/documents/synthetic/`).`
- `Document intake metadata and anonymized indexing tables.`

**Expected Outputs:**
- `Document inventory manifest: `data/interim/data_collection/document_inventory_manifest.json`.`
- `Document distribution statistics across required charity eligibility categories.`


### 1. Environment Setup & Library Imports

```python
import os
from pathlib import Path
import json
import pandas as pd
```

### 1. Setup Environment and Paths

```python
# Setup directory references
PROJECT_ROOT = Path("../../..")
DOCS_RAW_DIR = PROJECT_ROOT / "data" / "raw" / "documents"
OUTPUT_DIR = PROJECT_ROOT / "data" / "interim" / "data_collection"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print(f"Documents Directory: {DOCS_RAW_DIR.resolve()}")
```

### 2. Scan Supporting Document Files

```python
# Catalog available document files by extension
valid_exts = {'.pdf', '.jpg', '.jpeg', '.png', '.tiff'}
doc_records = []
for file_path in DOCS_RAW_DIR.rglob('*'):
    if file_path.is_file() and file_path.suffix.lower() in valid_exts:
        doc_records.append({
            'filename': file_path.name,
            'extension': file_path.suffix.lower(),
            'relative_path': str(file_path.relative_to(PROJECT_ROOT)),
            'size_kb': round(file_path.stat().st_size / 1024, 2)
        })
print(f"Total supporting documents cataloged: {len(doc_records)}")
```

### 3. Document Category Distribution Analysis

```python
# Summarize document types and sample volumes
df_docs = pd.DataFrame(doc_records)
if not df_docs.empty:
    print(df_docs['extension'].value_counts())
else:
    print("No document files present in raw directory (using template schema).")
```

### 4. Export Document Inventory Manifest

```python
# Persist catalog manifest
manifest_path = OUTPUT_DIR / 'document_inventory_manifest.json'
with open(manifest_path, 'w') as f:
    json.dump(doc_records, f, indent=2)
print(f"Document inventory manifest written to: {manifest_path}")
```
