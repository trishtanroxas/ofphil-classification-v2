### Description & Purpose
Validate YOLO format annotation files (.txt) for coordinate boundaries, label consistency, missing files, and zero-area bounding boxes.


### Research Objectives

1. Check that every image file has a corresponding annotation file (and vice-versa).
2. Validate that bounding box coordinates satisfy 0.0 <= x, y, w, h <= 1.0.
3. Detect overlapping or duplicate annotations within the same document region.
4. Produce an annotation verification summary report.


### Inputs & Expected Outputs

**Required Inputs:**
- `YOLO dataset labels directory (`data/processed/yolo_dataset/labels/`).`
- `YOLO `data.yaml` class mapping.`

**Expected Outputs:**
- `Annotation validation report (`data/interim/validated/yolo_annotation_audit.json`).`

### 1. Setup Paths and Label Scanner

```
PROJECT_ROOT = Path("../../../..")
LABEL_DIR = PROJECT_ROOT / "data" / "processed" / "yolo_dataset" / "labels"
print(f"Scanning annotation labels at: {LABEL_DIR.resolve()}")
```

### 2. Coordinate Invariant Verification

```
# Verify box coordinates are within [0, 1] range and width/height > 0
def validate_yolo_label_file(label_path, num_classes=6):
    errors = []
    with open(label_path, 'r') as f:
        for line_num, line in enumerate(f, 1):
            parts = line.strip().split()
            if len(parts) != 5:
                errors.append(f"Line {line_num}: invalid field count ({len(parts)})")
                continue
            cls_id, x, y, w, h = int(parts[0]), float(parts[1]), float(parts[2]), float(parts[3]), float(parts[4])
            if not (0 <= cls_id < num_classes):
                errors.append(f"Line {line_num}: invalid class_id {cls_id}")
            if not (0.0 <= x <= 1.0 and 0.0 <= y <= 1.0 and 0.0 < w <= 1.0 and 0.0 < h <= 1.0):
                errors.append(f"Line {line_num}: coordinate out of bounds")
    return errors
print("Label validation routine defined successfully.")
```
