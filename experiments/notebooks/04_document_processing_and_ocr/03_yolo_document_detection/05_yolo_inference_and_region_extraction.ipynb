### 2. ROI Cropping Function with Bounding Box Padding

```
def crop_detected_roi(image, bbox, padding_pct=0.02):
    # Crop image based on normalized YOLO bbox [x_center, y_center, w, h] with padding
    h_img, w_img = image.shape[:2]
    xc, yc, w, h = bbox
    x1 = max(0, int((xc - w/2 - padding_pct) * w_img))
    y1 = max(0, int((yc - h/2 - padding_pct) * h_img))
    x2 = min(w_img, int((xc + w/2 + padding_pct) * w_img))
    y2 = min(h_img, int((yc + h/2 + padding_pct) * h_img))
    return image[y1:y2, x1:x2]
print("ROI cropping utility configured with safety padding.")
```
