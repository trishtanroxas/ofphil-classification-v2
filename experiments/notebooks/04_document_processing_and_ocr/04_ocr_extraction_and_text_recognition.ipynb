### 1. Environment Setup & Library Imports

```
import json
from pathlib import Path
import re
```

### 1. Setup Environment and Extraction Paths

```
PROJECT_ROOT = Path("../../..")
OUTPUT_DIR = PROJECT_ROOT / "data" / "interim" / "extracted"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print("OCR Text Recognition & Parsing Engine.")
```

### 2. Regular Expression & Entity Extractors

```
# Define patterns for financial values, dates, and institutional terms
CURRENCY_PATTERN = re.compile(r'(?:PHP|Php|P|\u20B1)?\s*([0-9]{1,3}(?:,[0-9]{3})*(?:\.[0-9]{2})?)')
DATE_PATTERN = re.compile(r'\b(\d{1,2}[/-]\d{1,2}[/-]\d{2,4})\b')
print("Regular expression patterns compiled for financial and date entity extraction.")
```

### 3. Structured Parsing Pipeline

```
def parse_ocr_text_block(raw_text):
    return {
        'raw_text': raw_text,
        'detected_amounts': CURRENCY_PATTERN.findall(raw_text),
        'detected_dates': DATE_PATTERN.findall(raw_text)
    }
print("Entity parser initialized.")
```

### 4. Serialization of Extraction Template

```
with open(OUTPUT_DIR / 'ocr_extracted_entities.json', 'w') as f:
    json.dump({'status': 'template_ready'}, f, indent=2)
print("Extraction output destination verified.")
```
