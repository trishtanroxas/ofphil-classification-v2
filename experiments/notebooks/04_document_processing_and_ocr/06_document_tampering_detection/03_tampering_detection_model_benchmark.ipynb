### 1. Environment Setup & Model Dependencies

Import scikit-learn, PyTorch, and metric reporting utilities.


```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

ROOT_DIR = Path.cwd().resolve()
while ROOT_DIR.name != 'ofphil-classification-v2' and ROOT_DIR.parent != ROOT_DIR:
    ROOT_DIR = ROOT_DIR.parent
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Compute device selected: {device}')
```

### 2. Data Loading & Feature Extraction

Load training and validation records, converting document images into normalized feature representations.


```python
SPLITS_DIR = ROOT_DIR / 'data' / 'processed' / 'tampering' / 'splits'
df_train = pd.read_csv(SPLITS_DIR / 'train.csv') if (SPLITS_DIR / 'train.csv').exists() else pd.DataFrame()
df_test = pd.read_csv(SPLITS_DIR / 'test.csv') if (SPLITS_DIR / 'test.csv').exists() else pd.DataFrame()
print(f'Loaded {len(df_train)} train and {len(df_test)} test manifest entries.')
```
