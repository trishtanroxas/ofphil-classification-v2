### Description & Purpose
Train and evaluate an Explainable Boosting Machine (EBM from InterpretML) to establish an inherently glass-box, interpretable benchmark for charity eligibility classification.


### Research Objectives

1. Train an Explainable Boosting Classifier (generalized additive model with pairwise interaction terms GA2M).
2. Generate exact feature shape functions (term contributions) for each charity eligibility class.
3. Evaluate model accuracy and F1 score against complex gradient boosting models (XGBoost, CatBoost).
4. Assess clinical and social work auditability of the learned decision boundaries.
